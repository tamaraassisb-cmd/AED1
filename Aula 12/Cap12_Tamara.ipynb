**1) Escreva um programa que imprima os números de 1 a 10 usando while.**

```python
nota = float(input("Qual é a nota? "))

while nota < 0 or nota > 10:
  print("Nota inválida!")
  nota = float(input("Qual é a nota? "))

print("Nota válida!")
```

**10) Leia valores positivos até o usuário digitar -1. Exiba a soma, a quantidade e a média. O valor -1 não entra no cálculo.**

```python
numero = float(input("Digite um número positivo ou -1 para sair: "))
soma = 0
quantidade = 0
media = 0

while numero != -1:
  soma += numero
  quantidade += 1
  numero = float(input("Digite um número: "))

if quantidade > 0:
  media = soma / quantidade
  print(f'A soma é {soma:.2f}')
  print(f'A quantidade é {quantidade:.2f}')
  print(f'A média é {media:.2f}')
else:
  print("Não há valores para cálculo.")
```

**11) O computador sorteia um número entre 1 e 50. O usuário tenta adivinhar sem limite de tentativas. O programa informa se o número secreto é maior ou menor.**


```python
import random
numero_sorteado = random.randint(1,100)
acertou = False

while not acertou:
  chute = int(input("Chute um número de 1 a 100: "))

  if chute == numero_sorteado:
    print("Você acertou!")
    acertou = True
  elif chute < numero_sorteado:
    print("O número sorteado é maior.")
  else:
    print("O número sorteado é menor.")
```

**12) Adicione pontuação ao jogo. O jogador começa com 100 pontos e perde 10 pontos a cada erro. Há limite de 7 tentativas. Exiba a pontuação ao final.**


```python
import random

limite = 7
penalidade = 10

numero_sorteado = random.randint(1,100)
tentativas = 0
pontos = 100
acertou = False

while tentativas < limite and not acertou:
  restantes = limite - tentativas
  chute = int(input(f"Chute um número de 1 a 100. Você tem {restantes} tentativas restantes: "))
  tentativas += 1

  if chute == numero_sorteado:
    print("Você acertou!")
    acertou = True
  else:
    pontos -= penalidade
    if chute < numero_sorteado:
      print("O número sorteado é maior.")
    else:
      print("O número sorteado é menor.")

if acertou:
  print(f'Você acertou em {tentativas} tentativas!')
  print(f'Sua pontuação final é {pontos}')
else:
  print("Você perdeu!")
  print(f'O número sorteado era {numero_sorteado}')
  print(f'Sua pontuação final é {pontos}')

```
