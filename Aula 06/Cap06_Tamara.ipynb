**12) Uma loja dá 10% de desconto em compras de R$ 300 ou mais; sem atingir esse valor, não há desconto. Represente esse problema em quatro formas:**

(1) descrição em linguagem natural;

(2) fluxograma em Mermaid;

(3) pseudocódigo;

(4) prévia em Python.


1) linguagem natual

ler o valor da compra, se o valor for igual ou maior que R$ 300,00 dar 10% de desconto, se não nao há desconto. Calcular o valor de desconto, se houver, e mostrar o valor final.

image.png

ALGORITMO DESCONTO

INICIO

    LER valor_da_compra

      SE valor_da_compra >= 300 ENTÃO
      valor_da_compra = valor_da_compra * 0,9
      SENÃO
      valor_da_compra = valor_da_compra
      FIM_SE
    
    ESCREVA "Valor final é: ", valor_da_compra

FIM






```python
valor_da_compra = float(input("Digite o valor da compra: "))

if valor_da_compra >= 300:
  valor_da_compra = valor_da_compra * 0.9
else:
  valor_da_compra = valor_da_compra

print("Valor final é: ", valor_da_compra)
```
