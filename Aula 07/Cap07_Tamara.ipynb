**12) Colete o nome de um produto, seu preço original e um percentual de desconto. Calcule o valor do desconto em reais e o preço final, exibindo tudo formatado com duas casas decimais.**


```python
produto = input("Informe o nome do produto: ")
preco_original = float(input("Informe o preço do produto: "))
percentual_desconto = float(input("Informe o percentual de desconto: "))

desconto = preco_original * (percentual_desconto / 100)
preco_final = preco_original - desconto

print(f'O valor do desconto foi de R${desconto:.2f} e o preço final é de R${preco_final:.2f}')
```

**13) Colete o investimento em uma campanha de marketing, o número de leads gerados e o número de vendas realizadas. Calcule e exiba o custo por lead e a taxa de conversão.**

```python
investimento = float(input("Informe o investimento: "))
leads = int(input("Informe o número de leads: "))
vendas = int(input("Informe o número de vendas: "))

custo_por_lead = investimento / leads
taxa_conversao = vendas / leads

print(f'O custo por lead foi de R${custo_por_lead:.2f} e a taxa de conversão foi de {taxa_conversao:.2f}')
```
