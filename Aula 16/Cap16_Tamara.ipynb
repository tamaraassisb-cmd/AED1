**11) Refatore um programa com dados de alunos para usar quatro funções: calcular_media, classificar_aluno, criar_aluno e mostrar_relatorio. O código principal deve ter no máximo dez linhas.**

```python
def calcular_media(n1, n2, n3):
  return (n1 + n2 + n3) / 3

def classificar_aluno(media):
  if media >= 7:
    return 'Aprovado'
  elif media >= 5:
    return 'Recuperação'
  return 'Reprovado'

def criar_aluno(nome, n1, n2, n3):
  media = calcular_media(n1, n2, n3)
  situacao = classificar_aluno(media)
  return {
    'nome' : nome,
    'n1' : n1,
    'n2' : n2,
    'n3' : n3,
    'media' : media,
    'situacao': situacao
  }

def mostrar_relatorio(aluno):
  print(f'Aluno : {aluno["nome"]}')
  print(f'Média : {aluno["media"]:.2f}')
  print(f'Situação: {aluno["situacao"]}')
  print('-' * 30)

# Testes das funções de cálculo
assert calcular_media(10, 8, 6) == 8.0
assert classificar_aluno(7.0) == 'Aprovado'
assert classificar_aluno(5.0) == 'Recuperação'
assert classificar_aluno(4.99) == 'Reprovado'

aluno_teste = criar_aluno('Teste', 10, 8, 6)
assert aluno_teste['media'] == 8.0
assert aluno_teste['situacao'] == 'Aprovado'

# Código principal — limpo e curto
aluno = criar_aluno('Ana', 8.0, 7.5, 9.0)
mostrar_relatorio(aluno)
print('Testes passaram!')
```

**12) Crie estatisticas_vendas(vendas) que retorne três valores: total vendido, menor venda e maior venda. Se a lista estiver vazia, retornar 0.0, 0.0, 0.0.**

```python
def estatisticas_vendas(vendas):
 if not vendas:
  return 0.0, 0.0, 0.0
 total = sum(vendas)
 menor = min(vendas)
 maior = max(vendas)
 return total, menor, maior

total, menor, maior = estatisticas_vendas([80, 100, 95])
print(f'Total : {total}')
print(f'Menor : {menor}')
print(f'Maior : {maior}')
```
