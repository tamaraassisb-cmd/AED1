**1) Leia uma palavra digitada pelo usuário e exiba: o primeiro caractere, o último caractere e o tamanho total da palavra. Trate o caso em que o usuário pressiona Enter sem digitar nada.**

```python
palavra = input("Digite uma palavra: ") .strip()

if len(palavra) == 0:
  print("Você não digitou nenhuma palavra.")
else:
  print(f'O primeiro caractere é: {palavra[0]}')
  print(f'O último caractere é: {palavra[-1]}')
  print(f'O tamanho da palavra é: {len(palavra)}')
```

**4) Leia um CPF (formatado como 123.456.789-00 ou sem formatação como 12345678900), remova os caracteres de pontuação e verifique se o resultado tem exatamente 11 dígitos numéricos.**

```python
cpf_digitado = input("Digite o seu CPF: ") .strip()
cpf = cpf_digitado.replace('.','').replace('-','')

if cpf.isdigit() and len(cpf) == 11:
  print("CPF válido")
else:
  print("CPF inválido")
```

**5) Leia um endereço de e-mail e verifique se ele satisfaz quatro condições mínimas: exatamente um arroba, ponto após o arroba, ausência de espaços e comprimento mínimo de 5 caracteres.**


```python
email = input('e-mail: ') .strip() .lower()

# Condição 1: Tem que ter 1 @
um_arroba = email.count('@') == 1

pos_arroba = email.find('@') #localizar o @

#Condição 2: tem que ter ponto depois do @
if pos_arroba != -1:
  depois_do_arroba = email[pos_arroba + 1:]
  tem_ponto_depois = '.' in depois_do_arroba

#Condição 3: sem espaços
sem_espacos = ' ' not in email

#Condição 4: tamanho mínimo
tamanho_ok = len(email) >= 5

if um_arroba and tem_ponto_depois and sem_espacos and tamanho_ok:
  print('E-mail válido')
else:
  print('E-mail inválido')



```

**11) Leia um parágrafo e exiba seis estatísticas: número de caracteres com espaços, número de caracteres sem espaços, número de palavras, número de frases (contadas por pontuação final), número de vogais e a palavra mais longa.**


```python
texto = input("Digite um parágrafo: ")

#Caracteres
caracteres_com_espacos = len(texto)
caracteres_sem_espacos = len(texto.replace(' ',''))

#Palavras
palavras = texto.split()
numero_palavras = len(palavras)

#Frases
num_frases = texto.count('.') + texto.count('!') + texto.count('?')

#vogais
vogais = 'aeiouáéíóúàâêôãõ'
num_vogais = 0
for caracteres in texto.lower():
  if caracteres in vogais:
    num_vogais += 1

#Palavra mais longa
if numero_palavras > 0:
  maior_palavra = max(palavras, key=len)
else:
  maior_palavra = 'Não há'

print('Resultados')
print(40*'-')
print(f'Caracteres com espaços: {caracteres_com_espacos}')
print(f'Caracteres sem espaços: {caracteres_sem_espacos}')
print(f'Número de palavras: {numero_palavras}')
print(f'Número de frases: {num_frases}')
print(f'Número de vogais: {num_vogais}')
print(f'Palavra mais longa: {maior_palavra}')

```

**12) Leia uma linha no formato produto,categoria,preco,quantidade, processe cada campo e exiba o nome do produto, a categoria e o valor total (preço × quantidade).**

```python
linha = input('produto, categoria, preço, quantidade: ')
campos = linha.split(',') #divide a linha pela vírgula

if len(campos) != 4:
  print('Linha inválida')
  print('Digitar no formato: produto, categoria, preço, quantidade')
else:
  produto = campos[0].strip().title()
  categoria = campos[1].strip().title()

preco = float(campos[2].strip())
quantidade = int(campos[3].strip())

total = preco * quantidade

print(f'Produto: {produto}')
print(f'Categoria: {categoria}')
print(f'Valor total: {total}')
```
