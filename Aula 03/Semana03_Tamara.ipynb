## O que eu aprendi sobre bits
*Bit é um conceito/convenção para representar uma unidade de informação que possui dois estados possíveis.*

O bit não é o fenômeno físico.
O fenômeno físico é o jeito que encontramos de materializar/armazenar/transmitir essa informação.

Então, dependendo da tecnologia, um bit pode ser representado por carga elétrica, tensão elétrica, estado magnético, luz, etc.

Não existe uma lei da natureza dizendo que:

> 8 bits = 1 byte

O byte é uma unidade que foi definida pelos padrões da computação.

BIT   ->   0 ou 1    ->    8 bits agrupados     ->     1 BYTE      ->  256 combinações (2^8)

1KB = 1.024 bytes PORQUE 2^10 = 1024 (tem que ser um número de potência 2)



```python
# Calculadora de tamanho de imagem

largura = int(input("Digite a largura da imagem: "))
altura = int(input("Digite a altura da imagem:  "))
total_area_bytes = (largura * altura) * 3
print(total_area_bytes, "bytes")
total_area_kb = total_area_bytes / 1024
print(total_area_kb, "KB")
total_area_mb = total_area_kb / 1024
print(f'{total_area_mb:.1f} MB')
```

```python
# Calculadora de tamanho de áudio

tempo = float(input("Digite o tempo de áudio em minutos: "))
taxa = float(input("Digite a taxa de amostragem: "))
total_audio_bytes = ((tempo * 60) * taxa) / 8

resposta_estereo = input("O áudio é estéreo? (s/n): ")

if resposta_estereo == 's':
    total_audio_bytes = total_audio_bytes * 2
else:
    total_audio_bytes = total_audio_bytes * 1

print(total_audio_bytes, "bytes")
total_audio_kb = total_audio_bytes / 1024
print(total_audio_kb, "KB")
total_audio_mb = total_audio_kb / 1024
print(f'{total_audio_mb:.1f} MB')
```

```python
# Conversor binário

numero = int(input("Qual é o número? "))
print(f'{numero:3d} -> {numero:08b}')

letra = str(input("Qual é a letra? "))
print(f'{letra} -> {ord(letra):08b}')
```
