**10) Peça as medidas dos três lados de um triângulo. Primeiro, verifique se os três valores realmente formam um triângulo. Se formarem, classifique-o como equilátero, isósceles ou escaleno.**

```python
lado_a = float(input("Informe o valor do lado A: "))
lado_b = float(input("Informe o valor do lado B: "))
lado_c = float(input("Informe o valor do lado C: "))

if lado_a + lado_b > lado_c and lado_b + lado_c > lado_a and lado_a + lado_c > lado_b:
  if lado_a == lado_b == lado_c:
    print('O triângulo é equilátero.')
  elif lado_a == lado_b or lado_b == lado_c or lado_a == lado_c:
    print('O triângulo é escaleno.')
  else:
    print('O triângulo é isósceles.')
else:
  print('Não é um triângulo.')

```

**11) Peça o valor da compra e a distância de entrega em quilômetros. O frete é grátis quando a compra é maior que R$ 300 e a distância é de até 20 km. Em qualquer outra situação, o frete custa 8% do valor da compra mais RS 1,50 por quilômetro.**


```python
valor_compra = float(input("Qual é o valor da compra? "))
distancia = float(input("Qual é a distância da entrega em km? "))

if valor_compra > 300 and distancia <= 20:
  print("Frete grátis")
else:
  taxas = (valor_compra * 0.08) + (1.5 * distancia)
  valor_final = valor_compra + taxas
  print(f'O valor do frete é de R$ {taxas}')
  print(f'O valor da compra com o frete é de R$ {valor_final}')
```

**12) Peça o preço de venda e o custo de um produto. Calcule a margem percentual e classifique-a em três faixas: baixa (menor que 10%), adequada (de 10% a 30%) ou alta (maior que 30%).**


```python
preco_venda = float(input("Qual é o valor de venda do produto: "))
custo = float(input("Qual é o valor de custo do produto: "))

if preco_venda <= 0:
 print('Erro: o preço de venda precisa ser maior que zero.')
else:
  margem_percentual = ((preco_venda - custo) / custo) * 100

  if margem_percentual < 10:
    print('Baixa')
  elif margem_percentual >= 10 and margem_percentual <= 30:
    print('Adequada')
  else:
    print('Alta')
```

**13) Peça a uma IA para criar um programa que converta uma temperatura em Celsius para Fahrenheit e para Kelvin. Depois, avalie a resposta recebida usando os pontos de auditoria apresentados a seguir.**


```python
# 1. Solicitamos a temperatura em Celsius ao usuário
# Usamos 'float' para permitir que o usuário digite números decimais (ex: 25.5)
celsius = float(input("Digite a temperatura em Celsius (°C): "))

# 2. Aplicamos a fórmula de Celsius para Fahrenheit
# Fórmula: (°C * 9/5) + 32
fahrenheit = (celsius * 9 / 5) + 32

# 3. Aplicamos a fórmula de Celsius para Kelvin
# Fórmula: °C + 273.15
kelvin = celsius + 273.15

# 4. Exibimos os resultados formatados
print("\n--- Resultados da Conversão ---")
print(f"Temperatura em Celsius: {celsius}°C")
print(f"Temperatura em Fahrenheit: {fahrenheit:.2f}°F")
print(f"Temperatura em Kelvin: {kelvin:.2f}K")
```
