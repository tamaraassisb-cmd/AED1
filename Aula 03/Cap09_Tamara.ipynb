**3) Peça a temperatura e classifique-a em três faixas: abaixo de 15°C é frio, de 15°C a 25°C (sem incluir 25°C) é agradável, e 25°C ou mais é quente.**


```python
temperatura = float(input("Qual é a temperatura em ºC? "))

if temperatura < 15:
  print("Frio")
elif temperatura >= 15 and temperatura < 25:
  print("Agradável")
else:
  print("Quente")
```

**4) Peça o valor de uma compra. Se o valor for maior que R$ 200,00, aplique um desconto de 10%; caso contrário, o valor final é igual ao valor original. Exiba o valor original e o valor final, com duas casas decimais.**

```python
compra = float(input("Qual é o valor da compra? "))

if compra > 200:
  compra_final = compra - (compra * 0.1)
  print(f'O valor com desconto é de {compra_final:.2f}')
else:
  print(f'O valor final é de {compra:.2f}')
```

**10) Peça quatro notas (Prova 1, Prova 2 e Prova 3, com pesos 30%, 30% e 20%, e Projetos, com peso 20%) e calcule a média ponderada. Se a média for >= 7, exiba "Aprovado". Se for >= 5, exiba "Em recuperação", peça a nota da prova de recuperação, calcule a nova média (a média simples entre a média ponderada e a nota da recuperação) e informe se o aluno foi aprovado ou reprovado após a recuperação. Em qualquer outro caso, exiba "Reprovado".**


```python
p_1 = float(input("Qual é a nota da prova 1? "))
p_2 = float(input("Qual é a nota da prova 2? "))
p_3 = float(input("Qual é a nota da prova 3? "))

projetos = float(input("Qual é a nota dos projetos? "))

media_ponderada = (p_1 * 0.3) + (p_2 * 0.3) + (p_3 * 0.2) + (projetos * 0.2)

if media_ponderada >= 7:
  print(f'Média: {media_ponderada:.2f}')
  print("Aprovado")
elif media_ponderada >= 5:
  print(f'Média: {media_ponderada:.2f}')
  print("Em recuperação")
  nota_rec = float(input("Qual é a nota da recuperação? "))
  nova_media = (media_ponderada + nota_rec) / 2
  if nova_media >= 7:
    print(f'Média: {nova_media:.2f}')
    print("Aprovado")
  else:
    print(f'Média: {media_ponderada:.2f}')
    print("Reprovado")
else:
  print(f'Média: {media_ponderada:.2f}')
  print("Reprovado")
```

**11) Peça uma pontuação de risco entre 0 e 100 e classifique-a em três faixas: de 0 a 30 (incluindo 30) é baixo risco; de 31 a 70 (incluindo 70) é risco médio; acima de 70 é alto risco.**

```python
risco = float(input("Qual é a pontuação de risco? "))

if risco < 0 or risco > 100:
  print("Pontuação inválida")
elif risco <= 30:
  print("Baixo risco")
elif risco <= 70:
  print("Risco médio")
else:
  print("Alto risco")
```
