**5) Leia o valor de uma compra. Se o valor for maior ou igual a R$ 200,00, aplique 10% de desconto. Caso contrário, mantenha o valor original. Exiba o valor final.**

```python
valor_compra = float(input("Qual é o valor da compra? "))

if valor_compra >= 200:
  valor_final = valor_compra * 0.9
  print(f'O valor final é de R$ {valor_final:.2f}')
else:
  print(f'Desconto não aplicado, o valor final é de R$ {valor_compra:.2f}')
```

**10) Leia o valor de uma compra e a região de entrega. Se a compra for maior ou igual a RS 300,00, o frete é gratuito. Caso contrário: região "local" tem frete de RS 10,00; "estadual", RS 25,00; "nacional", RS 50,00. Exiba o valor da compra, o frete e o total final.**

```python
valor_compra = float(input("Qual é o valor da compra? "))
local = "Votorantim"
estado = "SP"

if valor_compra >= 300:
  print("O frete é gratuito!")
else:
  cidade_entrega = input("Qual é a cidade de entrega? ")
  estado_entrega = input("Qual é o estado de entrega (sigla)? ")
  if cidade_entrega == local and estado_entrega == estado:
    valor_total = valor_compra + 10
    print(f'O frete é de R$ 10,00')
  elif cidade_entrega != local and estado_entrega == estado:
    valor_total = valor_compra + 25
    print(f'O frete é de R$ 25,00')
  else:
    valor_total = valor_compra + 50
    print(f'O frete é de R$ 50,00')

  print(f'O valor final é de R$ {valor_total:.2f}')
```

**11) Leia o total gasto por um cliente no mês e classifique em: até RS 100, cliente eventual; acima de RS 100 e até RS 500, cliente regular; acima de RS 500, cliente prioritário. Se o cliente for prioritário, exiba também "Oferecer benefício especial".**

```python
valor_gasto = float(input("Qual é o valor gasto no mês? "))

if valor_gasto <= 100:
  print("Cliente eventual!")
elif valor_gasto > 100 and valor_gasto <= 500:
  print("Cliente regular!")
else:
  print("Cliente prioritário!")
  print("Oferecer benefício especial")
```
