**10) Peça a quantidade de alunos da turma e, para cada aluno, o nome e quatro notas (P1, P2, P3 e Projeto, com pesos 30%, 30%, 20% e 20%). Calcule a média ponderada de cada aluno, classifique-o (Aprovado se média >= 7, Recuperação se média >= 5, Reprovado nos demais casos), conte quantos alunos ficaram em cada situação, e ao final exiba a média geral da turma, o aluno com a maior média e o aluno com a menor média.**

```python
alunos = int(input('Digite a quantidade de alunos: '))
soma_medias = 0
aprovados = 0
recuperacao = 0
reprovados = 0

for i in range (1, alunos + 1):
  p1 = float(input("Qual é a nota da P1? "))
  p2 = float(input("Qual é a nota da P2? "))
  p3 = float(input("Qual é a nota da P3? "))
  pr = float(input("Qual é a nota do Projeto? "))
  media = (p1 * 0.3) + (p2 * 0.3) + (p3 * 0.2) + (pr * 0.2)
  soma_medias += media
  if media >= 7:
    situacao = 'Aprovado'
  elif media >= 5:
    situacao = 'Recuperação'
  else:
    situacao = 'Reprovado'
  print(f'O aluno {i} ficou com média {media:.2f} e está {situacao}.')

  if situacao == 'Aprovado':
    aprovados += 1
  elif situacao == 'Recuperação':
    recuperacao += 1
  else:
    reprovados += 1

media_geral = soma_medias / alunos
print(f'A média geral da turma foi {media_geral:.2f}.')
print(f'{aprovados} alunos estão aprovados.')
print(f'{recuperacao} alunos estão em recuperação.')
print(f'{reprovados} alunos estão reprovados.')
```

**11) Para 10 vendedores, peça o valor vendido por cada um. Ao final, exiba o total vendido, a média de vendas, quantos vendedores bateram a meta de R$ 5.000,00 e o percentual correspondente.**

```python
total_vendas = 0
meta = 5000
bateram_meta = 0
media_vendas = 0

for i in range (1,11):
  vendas = float(input(f'Digite o valor vendido pelo vendedor {i}: '))
  total_vendas += vendas
  if vendas >= meta:
    bateram_meta += 1

media_vendas = total_vendas / 10 #vendedores

print(f'O total vendido foi de R$ {total_vendas:.2f}')
print(f'A média de vendas foi de R$ {media_vendas:.2f}')
print(f'{bateram_meta} vendedores bateram a meta de R$ {meta:.2f}')
print(f'O percentual de vendedores que bateram a meta foi de {(bateram_meta / 10) * 100:.2f}%')
```

**12) Peça um número inteiro n e calcule n! (fatorial de n), que é o produto de todos os inteiros de 1 até n.**

```python
n = int(input('Digite o número: '))
if n < 0:
 print('Fatorial não existe para números negativos.')
else:
 resultado = 1
 for i in range(1, n + 1):
  resultado *= i
 print(f'{n}! = {resultado}')
```
