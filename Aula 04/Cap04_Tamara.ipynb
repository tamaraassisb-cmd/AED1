**13) Antes de usar qualquer IA, escreva o seu próprio algoritmo para "calcular o valor de uma compra com desconto". Depois, peça a uma IA que resolva o mesmo problema e compare as duas soluções usando os critérios abaixo.**

```python
# MEU CÓDIGO - Começa pelo valor original

valor_original = float(input("Qual o valor do produto? "))

# Agora é calculado o desconto

desconto = float(input("Qual o valor do desconto? (em %)"))

desconto_percentual = desconto / 100

valor_com_desconto = valor_original - (valor_original * desconto_percentual)

# O resultado final será a saída

print(f'O valor final é de R$ {valor_com_desconto:.2f}')
```

```python
# CÓDIGO DA IA - Calculadora de Valor de Compra com Desconto

# 1. Entrada de dados
# Usamos o float para permitir valores com centavos
valor_compra = float(input("Digite o valor total da compra (R$): "))
porcentagem_desconto = float(input("Digite a porcentagem de desconto (ex: 10 para 10%): "))

# 2. Processamento
valor_desconto = (valor_compra * porcentagem_desconto) / 100
valor_final = valor_compra - valor_desconto

# 3. Saída formatada com duas casas decimais
print("-" * 30)
print(f"Valor original: R$ {valor_compra:.2f}")
print(f"Desconto aplicado: R$ {valor_desconto:.2f}")
print(f"Valor final a pagar: R$ {valor_final:.2f}")
```
