**4) Usando a lista de produtos do Exercício 3, calcule o valor total do estoque multiplicando o preço pela quantidade de cada item e acumulando o resultado.**

```python
produtos = [
    {'nome': 'carregador', 'preco': 20.90, 'quantidade': 2},
    {'nome': 'mouse', 'preco': 49.90, 'quantidade': 1},
    {'nome': 'teclado', 'preco': 59.90, 'quantidade': 1}
]

total_estoque = 0
for produto in produtos:
    valor_item = produto['preco'] * produto['quantidade']
    total_estoque += valor_item
print(f'O valor total do estoque é R$ {total_estoque:.2f}')
```

**5) Dada uma lista de comentários, conte quantos mencionam "atendimento" sem diferenciar maiúsculas de minúsculas.**

```python
"O atendimento foi excelente" -> contém "atendimento"? Sim (minúscula)
"Gostei do preço" -> contém "atendimento"? Não
"ATENDIMENTO rápido e eficiente" -> contém "atendimento"? Sim (maiúscula)
"A entrega atrasou" -> contém "atendimento"? Não
"atendimento poderia melhorar" -> contém "atendimento"? Sim (minúscula)
```

```python
comentarios = [
    "O atendimento foi excelente",
    "Gostei do preço",
    "ATENDIMENTO rápido e eficiente",
    "A entrega atrasou",
    "atendimento poderia melhorar"
]

contador = 0

for comentario in comentarios:
  if 'atendimento' in comentario.lower():
    contador += 1

print(f'Foram encontrados {contador} comentários que mencionam "atendimento".')
```

**6) Dada uma lista de dicionários com nome e media, gere um ranking em ordem decrescente de média.**

```python
alunos = [
 {"nome": "Maria", "media": 9.5},
 {"nome": "Roberta", "media": 7.5},
 {"nome": "Raquel", "media": 10.0},
 {"nome": "João", "media": 5.5}
]

ranking = sorted(alunos, key=lambda aluno : aluno['media'], reverse=True)

for posicao, aluno in enumerate(ranking, start=1):
  print(f'{posicao}º lugar: {aluno["nome"]} com média {aluno["media"]}')
```
