A fotografia é formada por uma combinação de cores dentro de um pixel, que são o RED, GREEN e BLUE, comumente conhecido como RGB. Cada cor possui 256 combinações, ou seja, 8 bits. O computador não visualiza a imagem da mesma forma que o humano, ele enxerga os códigos binários que compõe os pixels. A resolução de uma imagem determina quantos pixels ela tem, por exemplo uma imagem de 1.000 x 1.080 possui 1.080.000 pixels. Cada arquivo possui a sua extensão e alguns podem comprimir os dados de armazenamento de uma imagem. A extensão JPEG pode reduzir alguns bits para diminuir o tamanho da imagem e com isso traz algumas perdas que podem ou não ser perceptíveis ao olho humano. Já o arquivo PNG encontra formas inteligentes de comprimir o arquivo preservando os dados originais.

# Exercício 15 — Prompt Contract

Use o Prompt Contract da seção 2.13 do capítulo para pedir a uma IA que explique como
letras se transformam em números na tabela ASCII. Depois, avalie se a resposta da IA
cumpriu cada parte do contrato.

**A resposta dada pela IA está simples e satisfatória.**

Claro! Vamos partir do zero, sem assumir que você já sabe programação.

1. O computador não entende diretamente a letra "A"

Para nós:

A é uma letra.

Mas o computador trabalha fundamentalmente com bits, ou seja, 0 e 1.

Então precisamos de uma maneira de dizer:

"Quando eu quiser representar a letra A, vou usar determinado número."

É aí que entra a tabela ASCII.

2. A tabela ASCII é como um "dicionário"

A ASCII estabelece uma correspondência entre caracteres e números.

Por exemplo:

Caractere	Número ASCII
A	65
B	66
C	67
a	97
b	98
0	48

Então, quando você escreve:

A

a tabela ASCII diz:

A → 65

O número 65 é o código ASCII da letra A.

3. Mas o computador ainda precisa transformar o 65 em bits

Aqui entra a parte que você acabou de estudar.

O número 65, em binário, é:

01000001

Então podemos pensar na sequência:

A
↓
65
↓
01000001

Ou seja:

letra → número → bits

É assim que o computador consegue armazenar a letra.

4. E por que a letra B é 66?

A tabela ASCII foi organizada de forma que as letras maiúsculas ficam em sequência:

A → 65
B → 66
C → 67
D → 68
...

Então existe uma relação numérica entre elas.

Por exemplo:

A = 65
B = 66

A diferença é simplesmente 1.

5. E o "a" minúsculo?

O a minúsculo possui outro código:

A → 65
a → 97

Isso acontece porque A e a são caracteres diferentes para a tabela ASCII.

Da mesma forma:

A ≠ a
6. E os números que você digita?

Aqui existe uma diferença MUITO importante:

O caractere:

"5"

também possui um código ASCII.

Ele é:

"5" → 53

Mas isso não significa que o número 5 seja armazenado como 53.

São coisas diferentes:

5 como valor numérico → número 5
"5" como caractere → código ASCII 53

Essa diferença entre número e caractere vai ser bastante importante quando você começar a programar.

🧠 Então guarde este caminho

Quando falamos de ASCII:

       CARACTERE
           ↓
     tabela ASCII
           ↓
         NÚMERO
           ↓
       representação
        em binário
           ↓
       BITS (0 e 1)

Por exemplo:

A → 65 → 01000001

E o mais importante: ASCII é uma tabela de correspondência. Ela não "transforma" a letra de forma matemática; ela estabelece que determinado caractere corresponde a determinado número.

# Projeto do capítulo — Meu nome em bits

O projeto reúne, em um único programa, tudo o que você praticou neste capítulo: percorrer
um texto letra por letra, obter o código decimal de cada letra com ord() e convertê-lo para
binário com format(). Pense nele como uma pequena demonstração de que você sabe
transformar texto em números — e números em bits.

```python
palavra = 'Tamara'

print('Letra | Decimal | Binario')
print('-' * 30)

for letra in palavra:
  codigo = ord(letra)
  binario = format(codigo, '08b')
  print(f'{letra:5} | {codigo:7} |  {binario}')
```
