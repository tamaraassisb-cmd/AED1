**4) Crie uma lista com os valores de vendas de cinco dias: [1200.50, 980.00, 1350.75, 1100.00, 1580.90]. Calcule o total vendido, a venda média, o maior valor e o menor valor.**

```python
vendas = [1200.50, 980.00, 1350.75, 1100.00, 1580.90]

total_vendido = sum(vendas)
media_vendas = total_vendido / len(vendas)
maior_valor = max(vendas)
menor_valor = min(vendas)

print(f'Total vendido: R$ {total_vendido:.2f}')
print(f'Venda média: R$ {media_vendas:.2f}')
print(f'Maior valor: R$ {maior_valor:.2f}')
print(f'Menor valor: R$ {menor_valor:.2f}')
```

**6) Crie uma lista com seis produtos. Cada produto deve ser um dicionário com nome, categoria e preco. Calcule: produto mais caro; produto mais barato; preço médio; produtos com preço acima da média; lista ordenada por preço crescente.**


```python
produtos = [
    {'nome': 'Mouse sem fio', 'categoria': 'Informática', 'preco': 49.90},
    {'nome': 'Teclado', 'categoria': 'Informática', 'preco': 79.90},
    {'nome': 'Monitor', 'categoria': 'Informática', 'preco': 129.90},
    {'nome': 'Impressora', 'categoria': 'Informática', 'preco': 199.90},
    {'nome': 'Microfone', 'categoria': 'Informática', 'preco': 99.90},
    {'nome': 'Webcam', 'categoria': 'Informática', 'preco': 79.90},
]
precos = []
for produto in produtos:
  precos.append(produto['preco'])

media = sum(precos) / len(precos)

mais_caro = max(produtos, key=lambda p: p['preco'])
mais_barato = min(produtos, key=lambda p: p['preco'])
print(f'Produto mais caro: {mais_caro["nome"]} | Preço: R$ {mais_caro["preco"]:.2f}')
print(f'Produto mais barato: {mais_barato["nome"]} | Preço: R$ {mais_barato["preco"]:.2f}')
print(f'Preço médio: R$ {media:.2f}')

acima_media = []
print('\nProdutos com preço acima da média:')
for produto in produtos:
  if produto['preco'] > media:
    print(f'{produto["nome"]}: R$ {produto["preco"]:.2f}')

# ordenar por preço crescente
produtos_ordenados = sorted(produtos, key=lambda p: p['preco'])
print('\nProdutos Ordenados por preço (menor para o maior):')
for produto in produtos_ordenados:
  print(f'{produto["nome"]}: R$ {produto["preco"]:.2f}')
```
